# OCR Medicine Matcher Benchmark

This notebook benchmarks fuzzy matching for noisy OCR-style medicine names.

It evaluates:
- Top-1 accuracy
- Top-3 accuracy
- Average confidence score
- Simple result visualization

```
import pandas as pd
import matplotlib.pyplot as plt
from rapidfuzz import fuzz, process
```

```
medicine_names = [
    "Paracetamol",
    "Azithromycin",
    "Amoxicillin",
    "Cetirizine",
    "Ibuprofen",
    "Metformin",
    "Aspirin",
    "Omeprazole",
    "Dolo 650",
    "Crocin"
]

ocr_samples = [
    ("Paracetamo1", "Paracetamol"),
    ("Azithromyc1n", "Azithromycin"),
    ("Amoxcillin", "Amoxicillin"),
    ("Cetirizin", "Cetirizine"),
    ("Ibuprofenn", "Ibuprofen"),
    ("Metfomin", "Metformin"),
    ("Asprin", "Aspirin"),
    ("Omeprazol", "Omeprazole"),
    ("Dolo650", "Dolo 650"),
    ("Crocim", "Crocin")
]
```

```
def find_matches(query, choices, limit=3):
    matches = process.extract(
        query,
        choices,
        scorer=fuzz.WRatio,
        limit=limit
    )
    return matches
```

```
results = []

for noisy_text, expected in ocr_samples:
    matches = find_matches(noisy_text, medicine_names, limit=3)
    top_predictions = [match[0] for match in matches]

    results.append({
        "OCR Text": noisy_text,
        "Expected": expected,
        "Top Match": top_predictions[0],
        "Top Score": matches[0][1],
        "Top-1 Correct": top_predictions[0] == expected,
        "Top-3 Correct": expected in top_predictions
    })

df = pd.DataFrame(results)
df
```

```
top1_accuracy = df["Top-1 Correct"].mean() * 100
top3_accuracy = df["Top-3 Correct"].mean() * 100
average_score = df["Top Score"].mean()

print(f"Top-1 Accuracy: {top1_accuracy:.2f}%")
print(f"Top-3 Accuracy: {top3_accuracy:.2f}%")
print(f"Average Confidence Score: {average_score:.2f}")
```

```
metrics = pd.DataFrame({
    "Metric": ["Top-1 Accuracy", "Top-3 Accuracy", "Average Confidence Score"],
    "Value": [top1_accuracy, top3_accuracy, average_score]
})

metrics
```

```
plt.figure(figsize=(7, 4))
plt.bar(metrics["Metric"], metrics["Value"])
plt.title("OCR Medicine Matching Benchmark")
plt.ylabel("Score")
plt.xticks(rotation=20)
plt.show()
```
